# A.  The `os` Module 

The `os` module in Python **provides a way to interact with the operating system—files, directories, environment variables, and more**.

Think of it as:
>**A bridge between your Python code and your computer’s file system.**

````python
# import the os module 
import os 
````

## Current Working Directory (CWD) 
The CWD is where our python script is running 


````python
cwd = os.getcwd()
print(f"Current Directory: {cwd}")
````

**👉 Why it matters:**

- All relative file paths depend on this
- Common source of bugs

## Changing the Directory 

```python
os.chdir("C:/Users/YourName/Documents")
print(os.getcwd()) # now that we have changed the directory, we have a new current working directory
```

⚠️ Be careful:

- If the path doesn’t exist → error
- Always verify before changing

## Listing Files and Directories 


````python
files = os.listdir()
print(files)
````

Specific path:
```python
files = os.listdir("C:/Users/YourName")
print(files)
```

## Creating and Removing Directories

Single directory:
```python
os.mkdir("new_folder") # make a directory is the cwd 
os.rmdir("new_folder") # removes the directory only if empty 

```

Nested directories:
```python
os.makedirs("parent/child/grandchild")
os.removedirs("parent/child/grandchild") # ⚠️ Only works if directories are empty.
```
